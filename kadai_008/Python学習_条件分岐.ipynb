```python
#課題：
#変数varが、3の倍数の場合は「Fizz」を出力
#変数varが、5の倍数の場合は「Buzz」を出力
#変数varが、3の倍数と5の倍数の両方に該当する場合は「FizzBuzz」を出力
#上記のどの場合にも該当しない場合は、変数varの値を出力
#変数varの代入値を51とする
var = 51
if var % 3 == 0:
  print("Fizz")
if var % 5 == 0:
  print("Buzz")
elif var % 3 == 0 and var % 5 == 0:
  print("FizzBuzz")
else:
  print(var)
```
