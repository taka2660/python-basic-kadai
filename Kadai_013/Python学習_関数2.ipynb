```python
def say_good_morning():
    print("おはようございます！")
    print("昨日はよく眠れましたか？")
    print("今日も一日頑張りましょう！")

def say_good_evening():
    print("こんばんは！")
    print("今日も一日お疲れさまでした。")

say_good_morning()
say_good_evening()
```

```python
def add_two_arguments(price, shipping_fee):
    # 与えられた引数priceと引数shipping_feeを加算し、変数totalに代入する
    total = price + shipping_fee

    # 変数$totalの値を出力する
    print(f"{total}円")

# 関数を呼び出し、引数として購入金額と送料を渡す
add_two_arguments(1200, 500)
```

```python
def tax_calc(price, tax_rate):
  #引数から消費税を計算してtaxに代入する
  total_price = price + (price * tax_rate)
  print(f"消費税は{total_price}円です。")
#関数を呼び出して引数として価格と税率を引き渡す
tax_calc(110, 0.1)
```
