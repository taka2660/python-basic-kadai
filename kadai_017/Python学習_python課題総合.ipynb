```python
#課題の内容
#名前(name)と年齢(age)の属性を持つHumanクラスを作成してください。
#Humanクラスには、以下の条件で標準出力(print)するcheck_adultメソッドを追加してください。
# ageが20以上の場合に大人であること
# そうでない場合に大人でないこと
#Humanクラスのインスタンスを複数生成してリストに追加し、リストの要素数分だけcheck_adultメソッドを呼び出してください。

#Humanクラスを定義
class Human:
  #nameとage属性とそれらを引数として受け取る指定を定義
  def __init__(self, name, age):
    self.name = name
    self.age = age
  #ageが20以上の場合に大人であることを判定するメソッド
  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}さんは{self.age}歳なので20歳以上の大人です。")
    else:
      print(f"{self.name}さんは{self.age}歳なので20歳の大人ではありません。")
#下記属性を設定してインスタン化する
human_a = Human("山本", 19)
human_b = Human("鈴木", 23)
human_c = Human("佐藤", 20)
#Humanクラスの各インスタンスをリストに代入する変数humansを設定して属性をリストに代入
humans = [human_a, human_b, human_c]
#human変数を定義してhumansで定義したリストを取り出す
for human in humans:
  human.check_adult()
```
