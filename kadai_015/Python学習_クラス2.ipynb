```python
class Human:  #Humanクラスを定義
  def __init__(self, name, age):
    #下記属性を定義
    self.name = ""
    self.age = ""
    #属性値を代入するメソッドを定義する
  def set_nameage(self, name, age):
      self.name = name
      self.age = age
    #値を出力するメソッドを定義
  def printinfo(self):
      print(self.name, self.age)
#インスタンス化インスタンス名をhumanとする
human = Human("太郎", "54")
#メソッドにアクセスして実行する
human.set_nameage("太郎", "54") #値を代入するメソッド
human.printinfo() #値を出力するメソッド
```
